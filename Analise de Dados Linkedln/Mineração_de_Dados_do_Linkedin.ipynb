## Número de conexões por periodo

```python
import datetime
import pandas as pd
import plotly.express as px
import numpy as np
```

```python
!pip install --upgrade plotly
```

```python
conexoes = pd.read_csv("/content/conexoes.csv")
conexoes.drop(labels = ["Unnamed: 0"], axis = 1, inplace = True)
```

```python
conexoes.head()
```

```python
def converter_data(data):
  return datetime.datetime.strptime(data, "%d %b %Y").strftime("%Y-%m-%d")
```

```python
 conexos["Connected On"] = conexoes["Connected On"].apply(converter_data)
```

```python
conexoes
```

```python
grafico = px.scatter(conexoes, x = "Full Name", y = "Connected On")
grafico.show()
```

```python
conexoes.groupby(by = "Connected On").count()
```

```python
grafico = px.line(conexoes.groupby(by = "Connected On").count(), title = "Novas conexões por data")
grafico.show()
```

```python
conexoes
```

```python
conexoes["Connected On"] = pd.to_datetime(conexoes["Connected On"])
```

```python
conexoes["ano"] = conexoes["Connected On"].dt.year
conexoes["mes"] = conexoes["Connected On"].dt.month
```

```python
grafico = px.line(conexoes.groupby(by = "ano").count(), title = "Novas conexões por ano")
grafico.show()
```

```python
grafico = px.line(conexoes.groupby(by = "mes").count(), title = "Novas conexões por mes")
grafico.show()
```

### Cargos

```python
np.unique(conexoes["Position"], return_counts = True)
```

```python
conexoes.shape
```

```python
len(np.unique(conexoes["Position"]))
```

```python
def altera_cargo(cargo):
  novo_cargo = cargo
  if cargo == "Desenvolvedora full stack":
    novo_cargo = "Desenvolvedor full stack"
  return novo_cargo
```

```python
conexoes["Position"] = conexoes["Position"].apply(altera_cargo)
```

```python
grafico = px.histogram(conexoes["Position"])
grafico.show()
```

```python
grafico = px.treemap(conexoes, path = ["Position", "Company", "Full Name"])
grafico.show()
```
