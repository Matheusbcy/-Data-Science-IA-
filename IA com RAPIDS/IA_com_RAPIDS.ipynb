# Com o Pandas (CPU)

```python
import pandas as pd

print(pd.__version__)
```

```python
df = pd.DataFrame()
df["id"] = [0, 1, 2, 2, 3, 3, 3]
df["val"] = [float(i + 10) for i in range(7)]
print(df)
```

```python
soma = df["val"].sum()
print(soma)
```

# Com o cuDF (GPU)

```python
import cudf
print(cudf.__version__)
```

```python
df = cudf.DataFrame()
df["id"] = [0, 1, 2, 2, 3, 3, 3]
df["val"] = [float(i + 10) for i in range(7)]
print(df)
```

```python
soma = df["val"].sum()
print(soma)
```

# Familiarizando com o cuDF

## Criação do objeto

```python
s = cudf.Series([1, 2, None, 3, 4])
print(s)
```

```python
df = cudf.DataFrame({
    "a": list(range(10)),
    "b": list(reversed(range(10))),
    "c": list(range(10))
})

df
```

```python
df_p = pd.DataFrame({"a": [0, 1, 2, 3], "b": [0.1, 0.2, None, 0.3]})
```

```python
# Converter um dataframe pandas em um dataframe cudf
df_c = cudf.DataFrame.from_pandas(df_p)
```
