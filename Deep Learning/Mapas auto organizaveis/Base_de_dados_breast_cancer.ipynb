```python
import minisom
import pandas as pd
import numpy as np
import sklearn
import matplotlib.pyplot as plt

from minisom import MiniSom
from sklearn.preprocessing import MinMaxScaler
from matplotlib.pylab import pcolor, colorbar, plot
```

```python
X = pd.read_csv("/content/entradas_breast.csv")
X
```

```python
y = pd.read_csv("/content/saidas_breast.csv")
y
```

```python
normalizador = MinMaxScaler(feature_range = (0, 1))
X = normalizador.fit_transform(X)
X
```

```python
type(y)
```

```python
y = np.array(y)
```

```python
type(y)
```

```python
som = MiniSom(x = 14, y = 14, input_len = X.shape[1], sigma = 1.0, learning_rate = 0.5, random_seed = 2)
```

```python
som.train_random(data = X, num_iteration = 6000)
```

```python
pcolor(som.distance_map().T)
colorbar() # MDI - mean inter neuron distance
```

```python
markers = ["o", "s"]
color = ["r", "g"]
```

```python
pcolor(som.distance_map().T)
colorbar()

for i, x in enumerate(X):
  w = som.winner(x)
  plot(w[0] + 0.5, w[1] + 0.5, markers[int(y[i])],
      markerfacecolor="None",
      markeredgecolor=color[int(y[i])],
      markeredgewidth=2, markersize=10)
```
