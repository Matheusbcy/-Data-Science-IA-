```python
import pandas as pd
import numpy as np
import tensorflow as tf
import sklearn
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras import utils as np_utils
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, accuracy_score
```

```python
base = pd.read_csv("/content/iris.csv")
base
```

```python
X = base.iloc[:, 0:4].values
y = base.iloc[:, 4].values
```

```python
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(y)
```

```python
y = np_utils.to_categorical(y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25)
```

```python
X_train.shape, y_train.shape
```

```python
X_test.shape, y_test.shape
```

```python
rede_neural = Sequential([
    tf.keras.layers.InputLayer(shape = (4, )),
    tf.keras.layers.Dense(units = 4, activation = "relu"),
    tf.keras.layers.Dense(units = 4, activation = "relu"),
    tf.keras.layers.Dense(units = 3, activation= "softmax")
])
```

```python
rede_neural.summary()
```

```python
rede_neural.compile(optimizer = "adam", loss = "categorical_crossentropy", metrics = ["categorical_accuracy"])
```

```python
rede_neural.fit(X_train, y_train, batch_size = 10, epochs = 1000)
```

```python
rede_neural.evaluate(X_test, y_test)
```

```python
prev = rede_neural.predict(X_test)
```

```python
prev = prev > 0.5
```

```python
y_test2 = [np.argmax(t) for t in y_test]
print(y_test2)
```

```python
prev2 = [np.argmax(t) for t in prev]
print(prev2)
```

```python
accuracy_score(y_test2, prev2)
```

```python
confusion_matrix(y_test2, prev2)
```

```python
import scikeras

from tensorflow.keras import backend as k
from sklearn.model_selection import cross_val_score
from scikeras.wrappers import KerasClassifier
```

```python
def criar_rede():
  k.clear_session()
  rede_neural = Sequential([
      tf.keras.layers.InputLayer(shape = (4, )),
      tf.keras.layers.Dense(units = 4, activation = "relu"),
      tf.keras.layers.Dense(units = 4, activation = "relu"),
      tf.keras.layers.Dense(units = 3, activation = "softmax"),
  ])
  rede_neural.compile(optimizer = "adam", loss = "categorical_crossentropy", metrics=["categorical_accuracy"])
  return rede_neural
```

```python
rede_neural = KerasClassifier(model = criar_rede, epochs = 250, batch_size = 10)
resultados = cross_val_score(estimator = rede_neural, X = X, y = y, cv = 10, scoring = "accuracy")
```

```python
resultados
```

```python
resultados.mean()
```

```python
resultados.std()
```

# Tuning dos parâmetros

```python
from sklearn.model_selection import GridSearchCV
```

```python
def criar_rede_dropout(optimizer, loss, kernel_initializer, activation, neurons):
  k.clear_session()
  model = Sequential([
      tf.keras.layers.InputLayer(shape = (4, )),
      tf.keras.layers.Dense(units = neurons, activation = activation, kernel_initializer=kernel_initializer),
      tf.keras.layers.Dropout(rate = 0.2),
      tf.keras.layers.Dense(units = neurons, activation = activation, kernel_initializer=kernel_initializer),
      tf.keras.layers.Dropout(rate = 0.2),
      tf.keras.layers.Dense(units = 3, activation = "softmax"),
  ])
  model.compile(optimizer = optimizer, loss = loss, metrics = ["categorical_accuracy"])
  return model
```

```python
rede_neural = KerasClassifier(model = criar_rede_dropout)
```

```python
params = {
    "batch_size": [10],
    "epochs": [250],
    "model__optimizer": ["adam"],
    "model__loss": ["categorical_crossentropy"],
    "model__kernel_initializer": ["glorot_uniform", "he_normal"],
    "model__activation": ["relu"],
    "model__neurons": [4],
}
```

```python
grid_search = GridSearchCV(estimator = rede_neural, param_grid = params, cv = 10)
```

```python
grid_search = grid_search.fit(X, y)
```

```python
best_params = grid_search.best_params_
```
